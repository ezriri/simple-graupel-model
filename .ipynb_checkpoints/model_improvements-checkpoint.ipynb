```python
PBASE = 705.        # pressure at cloud base
RD = 287.05 # gas constant for dry air (J/kg/K)
GAMMA = 0.0065 # constant lapse rate (ºC/m)
GRAV = 9.81 # gravity
P0 = 1013. # surface (sea level) pressure (mb)
T0 = 288. # surface temperature (K)

def ptz(p):
    """
    Calculate altitude at a given pressure. (Uses a constant lapse rate)

    Args:
        p (float): Pressure (mb).
    Returns:
        float: Altitude (m) at pressure p.
    """
    ex = (RD * GAMMA) / GRAV
    ptzr = T0 * (1.0 - (p / P0) ** ex) / GAMMA
    return ptzr
cloud_base_altitude = ptz(PBASE)
cloud_base_altitude
```

```python
plt.plot(ncfile['time'], ncfile_envi['cloud_top'], color='blue', label='Cloud Top', alpha=0.5)
plt.plot(ncfile['time'], ncfile_envi['thermal_base_1'], color='orange', label='Thermal Base1', alpha=0.5)
plt.plot(ncfile['time'], ncfile_envi['thermal_top_1'], color='red', label='Thermal Top1', alpha=0.5)
plt.plot(ncfile['time'], ncfile_envi['thermal_base_2'], color='green', label='Thermal Base2', alpha=0.5)
plt.plot(ncfile['time'], ncfile_envi['thermal_top_2'], color='purple', label='Thermal Top2', alpha=0.5)
# project cloud base
time = ncfile['time']
cloud_base = xr.DataArray(
    np.full(len(time), cloud_base_altitude/1000),
    coords={"time": time},
    dims="time",
    name="value")
plt.plot(time, cloud_base, color='gray', label='Cloud Base', alpha=0.5)
plt.ylabel("altitude (km)")
plt.xlabel("time (s)")
plt.legend()
plt.ylim(0,12)
```
